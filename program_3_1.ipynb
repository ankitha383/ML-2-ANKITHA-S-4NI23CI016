```python
import pandas as pd
df = pd.read_csv('testdata.csv')
df.head()

```

```python
X = df.iloc[:, :-1].values
Y = df.iloc[:, -1].values

hypothesis = None

for i in range(len(X)):
    if Y[i] == "yes":
        if hypothesis is None:
            hypothesis = X[i].copy()
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != X[i][j]:
                    hypothesis[j] = '?'

    print("Step", i + 1, "Hypothesis:", hypothesis)

print("\nFinal Hypothesis:", hypothesis)
```
